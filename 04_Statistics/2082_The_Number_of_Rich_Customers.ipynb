```python
import pandas as pd
from pandas.testing import assert_frame_equal
from decimal import Decimal, ROUND_HALF_UP

def percent(n,d):
    return float((Decimal(int(n))*100/Decimal(int(d))).quantize(Decimal('0.01'),rounding=ROUND_HALF_UP)) if d else None

store = pd.DataFrame([[1,1,600],[2,1,700],[3,2,500],[4,3,300],[5,3,300]], columns=["bill_id","customer_id","amount"])
expected = pd.DataFrame([[1]], columns=["rich_count"])
display(store)
display(expected)
```

## Solutions

### Solution 1 — Filter nunique
不是總消費大於 500。

```python
def solution_1(store):
    return pd.DataFrame({'rich_count':[store.loc[store.amount.gt(500),'customer_id'].nunique()]})

result_1 = solution_1(store.copy())
display(result_1)
```

### Solution 2 — Group maximum
先取每個 customer 最大單筆。

```python
def solution_2(store):
    return pd.DataFrame({'rich_count':[int(store.groupby('customer_id').amount.max().gt(500).sum())]})

result_2 = solution_2(store.copy())
display(result_2)
```

### Solution 3 — Group any
布林聚合可延伸多條件。

```python
def solution_3(store):
    g = store.assign(rich=store.amount.gt(500)).groupby('customer_id').rich.any()
    return pd.DataFrame({'rich_count':[int(g.sum())]})

result_3 = solution_3(store.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
