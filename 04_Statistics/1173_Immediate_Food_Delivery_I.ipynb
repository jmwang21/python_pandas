## Problem
Delivery(delivery_id,customer_id,order_date,customer_pref_delivery_date): report the percentage of all orders with identical order and preferred delivery dates, rounded to two decimals, as immediate_percentage.

## Practice first
Run the data cell and try the blank cell before reading solutions.

```python
import pandas as pd
from pandas.testing import assert_frame_equal
from decimal import Decimal, ROUND_HALF_UP

def percent(n,d):
    return float((Decimal(int(n))*100/Decimal(int(d))).quantize(Decimal('0.01'),rounding=ROUND_HALF_UP)) if d else None

delivery = pd.DataFrame([[1,1,"2024-01-01","2024-01-01"],[2,1,"2024-01-02","2024-01-03"],[3,2,"2024-01-01","2024-01-01"]], columns=["delivery_id","customer_id","order_date","customer_pref_delivery_date"])
expected = pd.DataFrame([[66.67]], columns=["immediate_percentage"])
display(delivery)
display(expected)
```

## Solutions

### Solution 1 — Boolean sum
分母是所有訂單；percent 使用精確 decimal 四捨五入。

```python
def solution_1(delivery):
    m = delivery.order_date.eq(delivery.customer_pref_delivery_date)
    return pd.DataFrame({'immediate_percentage':[percent(m.sum(),len(m))]})

result_1 = solution_1(delivery.copy())
display(result_1)
```

### Solution 2 — Value counts
get(True,0) 可處理沒有 immediate 的情形。

```python
def solution_2(delivery):
    counts = delivery.order_date.eq(delivery.customer_pref_delivery_date).value_counts()
    return pd.DataFrame({'immediate_percentage':[percent(counts.get(True,0),counts.sum())]})

result_2 = solution_2(delivery.copy())
display(result_2)
```

### Solution 3 — Query count
SQL-like filtering，再計算比例。

```python
def solution_3(delivery):
    n = len(delivery.query('order_date == customer_pref_delivery_date'))
    return pd.DataFrame({'immediate_percentage':[percent(n,len(delivery))]})

result_3 = solution_3(delivery.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
