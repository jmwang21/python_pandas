# 176. Second Highest Salary

## Problem Description

A payroll report requires the second-highest salary, independently of how many employees share a salary.

### Table Schema

**Employee**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| id | int | unique employee identifier (key) |
| salary | int | salary |

### Task

Find the second-largest distinct salary. Return a one-row DataFrame with column SecondHighestSalary; if a second distinct value does not exist, its value is null.

## Example

### Input

**employee**

| id | salary |
| --- | --- |
| 1 | 300 |
| 2 | 300 |
| 3 | 200 |
| 4 | 100 |

### Expected Output

| SecondHighestSalary |
| --- |
| 200 |

### Explanation

Although 300 occurs twice, it is the highest distinct salary only once. The second-highest distinct value is 200.

```python
import pandas as pd
from pandas.testing import assert_frame_equal
from decimal import Decimal, ROUND_HALF_UP

def percent(n,d):
    return float((Decimal(int(n))*100/Decimal(int(d))).quantize(Decimal('0.01'),rounding=ROUND_HALF_UP)) if d else None

employee = pd.DataFrame([[1,300],[2,300],[3,200],[4,100]], columns=["id","salary"])
expected = pd.DataFrame([[200]], columns=["SecondHighestSalary"])
display(employee)
display(expected)
```

## Solutions

### Solution 1 — Distinct nlargest
去重避免最高薪並列。

```python
def solution_1(employee):
    s = employee.salary.dropna().drop_duplicates().nlargest(2)
    return pd.DataFrame({'SecondHighestSalary':[s.iloc[1] if len(s)==2 else None]})

result_1 = solution_1(employee.copy())
display(result_1)
```

### Solution 2 — Exclude maximum
先排除最高薪再取 max。

```python
def solution_2(employee):
    s = employee.loc[employee.salary.lt(employee.salary.max()),'salary']
    return pd.DataFrame({'SecondHighestSalary':[s.max() if not s.empty else None]})

result_2 = solution_2(employee.copy())
display(result_2)
```

### Solution 3 — Dense rank
可延伸第 N 高。

```python
def solution_3(employee):
    s = employee.loc[employee.salary.rank(method='dense',ascending=False).eq(2),'salary']
    return pd.DataFrame({'SecondHighestSalary':[s.iloc[0] if len(s) else None]})

result_3 = solution_3(employee.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
