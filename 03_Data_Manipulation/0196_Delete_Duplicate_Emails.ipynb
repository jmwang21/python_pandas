# 196. Delete Duplicate Emails

## Problem Description

An address directory may contain multiple records with the same email address.

### Table Schema

**Person**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| id | int | unique person identifier (key) |
| email | varchar | email address |

### Task

For each email, retain only the record with the smallest id and delete all other records with that email. The Pandas solution must modify the original Person DataFrame in place. Returning a new DataFrame or assigning a new local variable alone does not satisfy the task. The remaining table may be in any order.

## Example

### Input

**person**

| id | email |
| --- | --- |
| 9 | a@x.com |
| 2 | b@x.com |
| 1 | a@x.com |
| 7 | b@x.com |

### Expected Output

| id | email |
| --- | --- |
| 1 | a@x.com |
| 2 | b@x.com |

### Explanation

For a@x.com keep id 1 instead of 9. For b@x.com keep id 2 instead of 7. The original DataFrame must end with those two records.

```python
import pandas as pd
from pandas.testing import assert_frame_equal
from decimal import Decimal, ROUND_HALF_UP

def percent(n,d):
    return float((Decimal(int(n))*100/Decimal(int(d))).quantize(Decimal('0.01'),rounding=ROUND_HALF_UP)) if d else None

person = pd.DataFrame([[9,"a@x.com"],[2,"b@x.com"],[1,"a@x.com"],[7,"b@x.com"]], columns=["id","email"])
expected = pd.DataFrame([[1,"a@x.com"],[2,"b@x.com"]], columns=["id","email"])
display(person)
display(expected)
```

## Solutions

### Solution 1 — Sort and deduplicate
必須原地修改，不可重新指派變數。

```python
def solution_1(person):
    person.sort_values('id',inplace=True)
    person.drop_duplicates('email',keep='first',inplace=True)

result_1 = person.copy()
solution_1(result_1)
display(result_1)
```

### Solution 2 — Transform
以 group minimum 選列；index 唯一。

```python
def solution_2(person):
    keep = person.id.eq(person.groupby('email').id.transform('min'))
    person.drop(index=person.index[~keep],inplace=True)

result_2 = person.copy()
solution_2(result_2)
display(result_2)
```

### Solution 3 — Idxmin
idxmin 回傳 index label；id 唯一。

```python
def solution_3(person):
    keep = person.groupby('email').id.idxmin()
    person.drop(index=person.index.difference(keep),inplace=True)

result_3 = person.copy()
solution_3(result_3)
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
