# 177. Nth Highest Salary

## Problem Description

A payroll database records employee salaries; multiple employees may earn the same amount.

### Table Schema

**Employee**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| id | int | unique employee identifier (key) |
| salary | int | salary |

### Task

Given an integer N, find the Nth largest distinct salary, counting equal salaries once. Return exactly one row with column getNthHighestSalary(N), replacing N with the actual argument. If there are fewer than N distinct salaries, return a null in that row, not an empty table. Treat nonpositive N as unavailable.

## Example

### Input

**employee**

| id | salary |
| --- | --- |
| 1 | 100 |
| 2 | 300 |
| 3 | 300 |
| 4 | 200 |

**Parameter:** N = 2

### Expected Output

| getNthHighestSalary(2) |
| --- |
| 200 |

### Explanation

For N = 2, the distinct salaries in descending order are 300, 200, 100. The second distinct salary is 200.

```python
import pandas as pd
from pandas.testing import assert_frame_equal
from decimal import Decimal, ROUND_HALF_UP

def percent(n,d):
    return float((Decimal(int(n))*100/Decimal(int(d))).quantize(Decimal('0.01'),rounding=ROUND_HALF_UP)) if d else None

employee = pd.DataFrame([[1,100],[2,300],[3,300],[4,200]], columns=["id","salary"])
N = 2
expected = pd.DataFrame([[200]], columns=["getNthHighestSalary(2)"])
display(employee)
display(expected)
```

## Solutions

### Solution 1 — Distinct sort
去重後再取位置。

```python
def solution_1(employee, N):
    s = employee.salary.dropna().drop_duplicates().sort_values(ascending=False)
    return pd.DataFrame({f'getNthHighestSalary({N})':[s.iloc[N-1] if 1<=N<=len(s) else None]})

result_1 = solution_1(employee.copy(), N)
display(result_1)
```

### Solution 2 — Dense rank
並列同順位且不跳號。

```python
def solution_2(employee, N):
    s = employee.loc[employee.salary.rank(method='dense',ascending=False).eq(N),'salary']
    return pd.DataFrame({f'getNthHighestSalary({N})':[s.iloc[0] if N>0 and len(s) else None]})

result_2 = solution_2(employee.copy(), N)
display(result_2)
```

### Solution 3 — Nlargest
只取需要的前 N 個。

```python
def solution_3(employee, N):
    s = employee.salary.dropna().drop_duplicates().nlargest(N) if N>0 else pd.Series(dtype=float)
    return pd.DataFrame({f'getNthHighestSalary({N})':[s.iloc[-1] if N>0 and len(s)==N else None]})

result_3 = solution_3(employee.copy(), N)
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
