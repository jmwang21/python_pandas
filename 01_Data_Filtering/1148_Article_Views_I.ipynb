# 1148. Article Views I

## Problem Description

An article-view log identifies the author, viewer and date of each view.

### Table Schema

**Views**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| article_id | int | article identifier |
| author_id | int | author identifier |
| viewer_id | int | viewer identifier |
| view_date | date | view date |

### Task

Find all authors who have viewed at least one article they authored (author_id equals viewer_id). Duplicate log records are possible. Return one row per qualifying author, with column id, sorted ascending.

## Example

### Input

**views**

| article_id | author_id | viewer_id | view_date |
| --- | --- | --- | --- |
| 1 | 3 | 3 | 2024-01-01 |
| 1 | 3 | 3 | 2024-01-01 |
| 2 | 1 | 1 | 2024-01-02 |
| 3 | 2 | 4 | 2024-01-03 |

### Expected Output

| id |
| --- |
| 1 |
| 3 |

### Explanation

Author 3 has two repeated self-views and author 1 has a self-view. Author 2's article was viewed by someone else.

```python
import pandas as pd
import re
from pandas.testing import assert_frame_equal
views = pd.DataFrame([[1,3,3,"2024-01-01"],[1,3,3,"2024-01-01"],[2,1,1,"2024-01-02"],[3,2,4,"2024-01-03"]], columns=["article_id","author_id","viewer_id","view_date"])
expected = pd.DataFrame([[1],[3]], columns=["id"])
display(views)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Filter and deduplicate
保留 DataFrame。

```python
def solution_1(views):
    return views.loc[views.author_id.eq(views.viewer_id),['author_id']].drop_duplicates().rename(columns={'author_id':'id'}).sort_values('id')

result_1 = solution_1(views.copy())
display(result_1)
```

### Solution 2 — Unique
unique 回傳陣列，需重新建表。

```python
def solution_2(views):
    return pd.DataFrame({'id':sorted(views.query('author_id == viewer_id').author_id.unique())})

result_2 = solution_2(views.copy())
display(result_2)
```

### Solution 3 — Group any
練習 group-level any；適合擴充統計。

```python
def solution_3(views):
    g = views.assign(self_view=views.author_id.eq(views.viewer_id)).groupby('author_id').self_view.any()
    return g[g].reset_index()[['author_id']].rename(columns={'author_id':'id'}).sort_values('id')

result_3 = solution_3(views.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
