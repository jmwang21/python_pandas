# 595. Big Countries

## Problem
World(name,continent,area,population,gdp): return name, population, area where area >= 3000000 OR population >= 25000000. Any order.

[Original problem and schema (Doocs)](https://github.com/doocs/leetcode/blob/main/solution/0500-0599/0595.Big%20Countries/README_EN.md)

## Practice first
Run the data cell and write your own answer before opening the solutions.

```python
import pandas as pd
import numpy as np
import re
from pandas.testing import assert_frame_equal
world = pd.DataFrame([["A","Asia",3000000,1,1],["B","Asia",1,25000000,1],["C","Asia",3000000,25000000,1],["D","Asia",1,1,1]], columns=["name","continent","area","population","gdp"])
expected = pd.DataFrame([["A",1,3000000],["B",25000000,1],["C",25000000,3000000]], columns=["name","population","area"])
display(world)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Boolean mask
直觀且向量化；注意括號及等號。

```python
def solution_1(world):
    return world.loc[(world.area >= 3000000) | (world.population >= 25000000), ['name','population','area']]

result_1 = solution_1(world.copy())
display(result_1)
```

### Solution 2 — Query
適合複雜條件的可讀性。

```python
def solution_2(world):
    return world.query('area >= 3000000 or population >= 25000000')[['name','population','area']]

result_2 = solution_2(world.copy())
display(result_2)
```

### Solution 3 — Union
集合聯集；必須去除同時符合條件的國家。

```python
def solution_3(world):
    return pd.concat([world[world.area >= 3000000],world[world.population >= 25000000]]).drop_duplicates('name')[['name','population','area']]

result_3 = solution_3(world.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
