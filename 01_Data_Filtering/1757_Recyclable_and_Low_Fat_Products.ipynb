Products(product_id,low_fats,recyclable): select product_id where both flags are Y. Any order.

```python
import pandas as pd
import re
from pandas.testing import assert_frame_equal
products = pd.DataFrame([[1,"Y","Y"],[2,"Y","N"],[3,"N","Y"]], columns=["product_id","low_fats","recyclable"])
expected = pd.DataFrame([[1]], columns=["product_id"])
display(products)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Boolean mask
兩條件最直接。

```python
def solution_1(products):
    return products.loc[products.low_fats.eq('Y') & products.recyclable.eq('Y'), ['product_id']]

result_1 = solution_1(products.copy())
display(result_1)
```

### Solution 2 — Row-wise all
可擴充多欄位條件。

```python
def solution_2(products):
    return products.loc[products[['low_fats','recyclable']].eq('Y').all(axis=1), ['product_id']]

result_2 = solution_2(products.copy())
display(result_2)
```

### Solution 3 — Query
SQL-like 條件。

```python
def solution_3(products):
    return products.query("low_fats == 'Y' and recyclable == 'Y'")[['product_id']]

result_3 = solution_3(products.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
