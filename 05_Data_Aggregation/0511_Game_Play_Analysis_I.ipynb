Activity(player_id,device_id,event_date,games_played): find each player's earliest login date, including zero games played. Return player_id,first_login.

```python
import pandas as pd
from pandas.testing import assert_frame_equal

activity = pd.DataFrame([[1,2,"2024-01-03",1],[1,3,"2024-01-01",0],[2,2,"2024-01-02",2]], columns=["player_id","device_id","event_date","games_played"])
expected = pd.DataFrame([[1,"2024-01-01"],[2,"2024-01-02"]], columns=["player_id","first_login"])
display(activity)
display(expected)
```

## Solutions

### Solution 1 — Group minimum
named aggregation 可直接指定輸出名稱。

```python
def solution_1(activity):
    return activity.groupby('player_id',as_index=False).agg(first_login=('event_date','min'))

result_1 = solution_1(activity.copy())
display(result_1)
```

### Solution 2 — Sort and deduplicate
可延伸保留首次登入的其他欄位。

```python
def solution_2(activity):
    return activity.sort_values('event_date').drop_duplicates('player_id')[['player_id','event_date']].rename(columns={'event_date':'first_login'})

result_2 = solution_2(activity.copy())
display(result_2)
```

### Solution 3 — Transform filter
依原粒度選擇 row；題目保證 player_id,event_date 唯一。

```python
def solution_3(activity):
    m = activity.event_date.eq(activity.groupby('player_id').event_date.transform('min'))
    return activity.loc[m,['player_id','event_date']].rename(columns={'event_date':'first_login'})

result_3 = solution_3(activity.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
