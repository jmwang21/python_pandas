## Practice first
Run the data cell and try the blank cell before reading solutions.

```python
import pandas as pd
from pandas.testing import assert_frame_equal

teacher = pd.DataFrame([[1,10,1],[1,10,2],[1,20,1],[2,10,1]], columns=["teacher_id","subject_id","dept_id"])
expected = pd.DataFrame([[1,2],[2,1]], columns=["teacher_id","cnt"])
display(teacher)
display(expected)
```

## Solutions

### Solution 1 — Nunique
最直接的 distinct count。

```python
def solution_1(teacher):
    return teacher.groupby('teacher_id',as_index=False).agg(cnt=('subject_id','nunique'))

result_1 = solution_1(teacher.copy())
display(result_1)
```

### Solution 2 — Deduplicate then size
去重的鍵不含 dept_id。

```python
def solution_2(teacher):
    return teacher.drop_duplicates(['teacher_id','subject_id']).groupby('teacher_id').size().reset_index(name='cnt')

result_2 = solution_2(teacher.copy())
display(result_2)
```

### Solution 3 — Distinct pairs value_counts
value_counts 可直接統計已去重的 teacher_id。

```python
def solution_3(teacher):
    return teacher[['teacher_id','subject_id']].drop_duplicates().teacher_id.value_counts().rename_axis('teacher_id').reset_index(name='cnt')

result_3 = solution_3(teacher.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
