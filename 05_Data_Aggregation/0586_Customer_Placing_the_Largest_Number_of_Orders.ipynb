# 586. Customer Placing the Largest Number of Orders

## Problem Description

An order database records the customer responsible for every order.

### Table Schema

**Orders**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| order_number | int | unique order identifier (key) |
| customer_number | int | customer identifier |

### Task

Count the orders placed by each customer_number, then identify the customer with the largest count. The problem guarantees exactly one customer has strictly more orders than every other customer. Return a one-row DataFrame with only customer_number.

## Example

### Input

**orders**

| order_number | customer_number |
| --- | --- |
| 1 | 1 |
| 2 | 2 |
| 3 | 2 |
| 4 | 3 |
| 5 | 2 |

### Expected Output

| customer_number |
| --- |
| 2 |

### Explanation

Customer 2 placed three orders, while customers 1 and 3 placed one each. The answer is customer 2.

```python
import pandas as pd
from pandas.testing import assert_frame_equal

orders = pd.DataFrame([[1,1],[2,2],[3,2],[4,3],[5,2]], columns=["order_number","customer_number"])
expected = pd.DataFrame([[2]], columns=["customer_number"])
display(orders)
display(expected)
```

## Solutions

### Solution 1 — Value counts
value_counts 預設依頻數降序。

```python
def solution_1(orders):
    return orders.customer_number.value_counts().head(1).rename_axis('customer_number').reset_index()[['customer_number']]

result_1 = solution_1(orders.copy())
display(result_1)
```

### Solution 2 — Group size and maximum
可自然延伸至並列第一。

```python
def solution_2(orders):
    g = orders.groupby('customer_number').size()
    return g[g.eq(g.max())].reset_index()[['customer_number']]

result_2 = solution_2(orders.copy())
display(result_2)
```

### Solution 3 — Mode
mode 回傳所有眾數；題目保證唯一。

```python
def solution_3(orders):
    return orders.customer_number.mode().to_frame(name='customer_number')

result_3 = solution_3(orders.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
