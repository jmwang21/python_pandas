# 596. Classes With at Least 5 Students

## Problem
Courses(student,class): return classes with at least five distinct students. Any order.

## Practice first

```python
import pandas as pd
from pandas.testing import assert_frame_equal

courses = pd.DataFrame([["A","Math"],["B","Math"],["C","Math"],["D","Math"],["E","Math"],["A","Art"],["B","Art"]], columns=["student","class"])
expected = pd.DataFrame([["Math"]], columns=["class"])
display(courses)
display(expected)
```

## Solutions

### Solution 1 — Group nunique
HAVING-style 聚合後篩選。

```python
def solution_1(courses):
    g = courses.groupby('class').student.nunique()
    return g[g.ge(5)].reset_index()[['class']]

result_1 = solution_1(courses.copy())
display(result_1)
```

### Solution 2 — Value counts
以一欄快速計數。

```python
def solution_2(courses):
    g = courses.drop_duplicates(['student','class'])['class'].value_counts()
    return g[g.ge(5)].rename_axis('class').reset_index()[['class']]

result_2 = solution_2(courses.copy())
display(result_2)
```

### Solution 3 — Group transform
可保留符合門檻的原始 rows，再選 distinct class。

```python
def solution_3(courses):
    s = courses.groupby('class').student.transform('nunique')
    return courses.loc[s.ge(5),['class']].drop_duplicates()

result_3 = solution_3(courses.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
