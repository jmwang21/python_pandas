Tweets(tweet_id,content): return IDs of tweets longer than 15 characters. Any order.

```python
import pandas as pd
import re
from pandas.testing import assert_frame_equal
tweets = pd.DataFrame([[1,"123456789012345"],[2,"1234567890123456"],[3,"hi"]], columns=["tweet_id","content"])
expected = pd.DataFrame([[2]], columns=["tweet_id"])
display(tweets)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Vectorized length
15 字元仍合法。

```python
def solution_1(tweets):
    return tweets.loc[tweets.content.str.len().gt(15),['tweet_id']]

result_1 = solution_1(tweets.copy())
display(result_1)
```

### Solution 2 — Map len
假設 content 非 null；Python function mapping。

```python
def solution_2(tweets):
    return tweets.loc[tweets.content.map(len)>15,['tweet_id']]

result_2 = solution_2(tweets.copy())
display(result_2)
```

### Solution 3 — Regex length
練習 regex 長度量詞；本題 str.len 較直觀。

```python
def solution_3(tweets):
    return tweets.loc[tweets.content.str.contains(r'(?s)^.{16}',na=False),['tweet_id']]

result_3 = solution_3(tweets.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
