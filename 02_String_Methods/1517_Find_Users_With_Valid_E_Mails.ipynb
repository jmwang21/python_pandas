# 1517. Find Users With Valid E-Mails

## Problem Description

A website must identify accounts whose registered email addresses follow its naming convention.

### Table Schema

**Users**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| user_id | int | unique user identifier (key) |
| name | varchar | user name |
| mail | varchar | registered email address |

### Task

Return all three original columns for users whose email starts with an ASCII letter (A–Z or a–z), followed by zero or more ASCII letters, digits, underscores, periods or hyphens, followed by exactly @leetcode.com in lowercase. Reject other domains, extra characters and invalid prefixes. Any row order is accepted.

## Example

### Input

**users**

| user_id | name | mail |
| --- | --- | --- |
| 1 | A | a-1@leetcode.com |
| 2 | B | 1a@leetcode.com |
| 3 | C | c@leetcodeXcom |
| 4 | D | D@leetcode.com |
| 5 | E | e@LEETCODE.COM |
| 6 | F | f@leetcode.com\n |

### Expected Output

| user_id | name | mail |
| --- | --- | --- |
| 1 | A | a-1@leetcode.com |
| 4 | D | D@leetcode.com |

### Explanation

a-1@leetcode.com and D@leetcode.com qualify. Leading digits, an incorrect domain separator, uppercase domain and a trailing newline fail.

```python
import pandas as pd
import re
from pandas.testing import assert_frame_equal
users = pd.DataFrame([[1,"A","a-1@leetcode.com"],[2,"B","1a@leetcode.com"],[3,"C","c@leetcodeXcom"],[4,"D","D@leetcode.com"],[5,"E","e@LEETCODE.COM"],[6,"F","f@leetcode.com\n"]], columns=["user_id","name","mail"])
expected = pd.DataFrame([[1,"A","a-1@leetcode.com"],[4,"D","D@leetcode.com"]], columns=["user_id","name","mail"])
display(users)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Regex fullmatch
fullmatch 避免尾端換行；dot 要跳脫。

```python
def solution_1(users):
    return users.loc[users.mail.str.fullmatch(r'[A-Za-z][A-Za-z0-9_.-]*@leetcode\.com',na=False)]

result_1 = solution_1(users.copy())
display(result_1)
```

### Solution 2 — Partition and validate
分段驗證方便定位錯誤。

```python
def solution_2(users):
    p = users.mail.str.partition('@')
    m = p[0].str.fullmatch(r'[A-Za-z][A-Za-z0-9_.-]*',na=False) & p[1].eq('@') & p[2].eq('leetcode.com')
    return users.loc[m]

result_2 = solution_2(users.copy())
display(result_2)
```

### Solution 3 — Python regex mapping
自訂驗證規則；通常向量化 API 較簡潔。

```python
def solution_3(users):
    pattern = re.compile(r'[A-Za-z][A-Za-z0-9_.-]*@leetcode\.com')
    return users.loc[users.mail.map(lambda x: isinstance(x,str) and bool(pattern.fullmatch(x)))]

result_3 = solution_3(users.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
