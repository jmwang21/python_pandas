# 1527. Patients With a Condition

## Problem Description

A hospital stores patient diagnoses as a space-delimited list of condition codes.

### Table Schema

**Patients**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| patient_id | int | unique patient identifier (key) |
| patient_name | varchar | name |
| conditions | varchar | zero or more space-separated codes |

### Task

Identify patients with at least one condition code beginning with DIAB1, representing type 1 diabetes. Match the beginning of a code, not an arbitrary substring of a longer code. Return patient_id, patient_name and conditions. Any row order is accepted.

## Example

### Input

**patients**

| patient_id | patient_name | conditions |
| --- | --- | --- |
| 1 | A | DIAB100 |
| 2 | B | ACNE DIAB123 |
| 3 | C | XDIAB100 |
| 4 | D |  |
| 5 | E | DIAB2 |

### Expected Output

| patient_id | patient_name | conditions |
| --- | --- | --- |
| 1 | A | DIAB100 |
| 2 | B | ACNE DIAB123 |

### Explanation

DIAB100 and ACNE DIAB123 contain qualifying codes. XDIAB100 does not start with DIAB1; an empty list and DIAB2 also fail.

```python
import pandas as pd
import re
from pandas.testing import assert_frame_equal
patients = pd.DataFrame([[1,"A","DIAB100"],[2,"B","ACNE DIAB123"],[3,"C","XDIAB100"],[4,"D",""],[5,"E","DIAB2"]], columns=["patient_id","patient_name","conditions"])
expected = pd.DataFrame([[1,"A","DIAB100"],[2,"B","ACNE DIAB123"]], columns=["patient_id","patient_name","conditions"])
display(patients)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Regex token start
不能用單純 substring，XDIAB100 不符合。

```python
def solution_1(patients):
    return patients.loc[patients.conditions.str.contains(r'(?:^| )DIAB1',na=False)]

result_1 = solution_1(patients.copy())
display(result_1)
```

### Solution 2 — Split and any
token-level 判斷。

```python
def solution_2(patients):
    m = patients.conditions.str.split().map(lambda codes: any(c.startswith('DIAB1') for c in codes))
    return patients.loc[m]

result_2 = solution_2(patients.copy())
display(result_2)
```

### Solution 3 — Explode
長表處理可擴充條件代碼分析。

```python
def solution_3(patients):
    s = patients.conditions.str.split(' ').explode()
    ids = s.index[s.str.startswith('DIAB1',na=False)]
    return patients.loc[patients.index.isin(ids)]

result_3 = solution_3(patients.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
