# 1667. Fix Names in a Table

## Problem
Users(user_id,name): make initial letter uppercase and remaining letters lowercase; return rows sorted by user_id.

[Original problem and schema (Doocs)](https://github.com/doocs/leetcode/blob/main/solution/1600-1699/1667.Fix%20Names%20in%20a%20Table/README_EN.md)

## Practice first
Run the data cell and write your own answer before opening the solutions.

```python
import pandas as pd
import numpy as np
import re
from pandas.testing import assert_frame_equal
users = pd.DataFrame([[2,"bOB"],[1,"ALIce"],[3,"x"]], columns=["user_id","name"])
expected = pd.DataFrame([[1,"Alice"],[2,"Bob"],[3,"X"]], columns=["user_id","name"])
display(users)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Capitalize
最直接；title 會大寫每個單字。

```python
def solution_1(users):
    return users.assign(name=users.name.str.capitalize()).sort_values('user_id')

result_1 = solution_1(users.copy())
display(result_1)
```

### Solution 2 — String slicing
練習向量化切片。

```python
def solution_2(users):
    return users.assign(name=users.name.str[:1].str.upper()+users.name.str[1:].str.lower()).sort_values('user_id')

result_2 = solution_2(users.copy())
display(result_2)
```

### Solution 3 — Map
Python function mapping；假設非 null。

```python
def solution_3(users):
    return users.assign(name=users.name.map(str.capitalize)).sort_values('user_id')

result_3 = solution_3(users.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
