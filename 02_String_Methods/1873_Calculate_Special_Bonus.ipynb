# 1873. Calculate Special Bonus

## Problem
Employees(employee_id,name,salary): bonus is salary for odd IDs whose names do not start with uppercase M, otherwise zero. Return employee_id,bonus sorted by ID.

[Original problem and schema (Doocs)](https://github.com/doocs/leetcode/blob/main/solution/1800-1899/1873.Calculate%20Special%20Bonus/README_EN.md)

## Practice first
Run the data cell and write your own answer before opening the solutions.

```python
import pandas as pd
import numpy as np
import re
from pandas.testing import assert_frame_equal
employees = pd.DataFrame([[3,"Ada",100],[2,"Ben",200],[1,"Mia",300],[5,"max",400]], columns=["employee_id","name","salary"])
expected = pd.DataFrame([[1,0],[2,0],[3,100],[5,400]], columns=["employee_id","bonus"])
display(employees)
display(expected)
```

## Solutions
Compare the approaches using the same input.

### Solution 1 — Where
where 保留 True 的資料。

```python
def solution_1(employees):
    m = employees.employee_id.mod(2).eq(1) & ~employees.name.str.startswith('M')
    return employees.assign(bonus=employees.salary.where(m,0))[['employee_id','bonus']].sort_values('employee_id')

result_1 = solution_1(employees.copy())
display(result_1)
```

### Solution 2 — NumPy where
np.where 回傳陣列。

```python
def solution_2(employees):
    m = (employees.employee_id%2==1) & ~employees.name.str.startswith('M')
    return employees.assign(bonus=np.where(m,employees.salary,0))[['employee_id','bonus']].sort_values('employee_id')

result_2 = solution_2(employees.copy())
display(result_2)
```

### Solution 3 — Conditional assignment
loc 條件賦值；注意 index 對齊。

```python
def solution_3(employees):
    out = employees[['employee_id']].assign(bonus=0)
    m = (employees.employee_id%2==1) & ~employees.name.str.startswith('M')
    out.loc[m,'bonus'] = employees.loc[m,'salary']
    return out.sort_values('employee_id')

result_3 = solution_3(employees.copy())
display(result_3)
```

## Check all solutions
Normalize row order for these examples.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
