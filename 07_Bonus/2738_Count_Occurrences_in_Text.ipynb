# 2738. Count Occurrences in Text

## Problem Description

A text corpus stores a unique file name and the contents of each file.

### Table Schema

**Files**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| file_name | varchar | unique file name (key) |
| content | text | file contents |

### Task

For each of the two lowercase words bull and bear, count the number of distinct files containing at least one occurrence surrounded on both sides by literal spaces. A word at the very start or end without a surrounding space, next to punctuation, or embedded in another word is not a match. Multiple valid occurrences in the same file count once for that word. Return word, count, with exactly two rows, including zero when absent. Any order is accepted.

## Example

### Input

**files**

| file_name | content |
| --- | --- |
| a | a bull bull market |
| b | bear market bull. |
| c | a bear and bull market |
| d | a bears or bullet story |

### Expected Output

| word | count |
| --- | --- |
| bull | 2 |
| bear | 1 |

### Explanation

File a contains bull twice but counts once. File b has bear at the start and bull followed by punctuation, so neither qualifies. File c contains both valid words. File d contains only longer words. The counts are bull=2 and bear=1.

```python
import pandas as pd
from pandas.testing import assert_frame_equal

files = pd.DataFrame([["a","a bull bull market"],["b","bear market bull."],["c","a bear and bull market"],["d","a bears or bullet story"]], columns=["file_name","content"])
expected = pd.DataFrame([["bull",2],["bear",1]], columns=["word","count"])
display(files)
display(expected)
```

## Solutions

### Solution 1 — Literal contains
空白為必要條件；每個檔案只算一次。

```python
def solution_1(files):
    return pd.DataFrame({'word':['bull','bear'],'count':[files.content.str.contains(' bull ',regex=False,na=False).sum(),files.content.str.contains(' bear ',regex=False,na=False).sum()]})

result_1 = solution_1(files.copy())
display(result_1)
```

### Solution 2 — Regex extractall and distinct file
lookaround 可匹配連續相鄰單字；先去重 file_name,word。

```python
def solution_2(files):
    m = files.set_index('file_name').content.str.extractall(r'(?<= )(bull|bear)(?= )').rename(columns={0:'word'}).reset_index()
    g = m.drop_duplicates(['file_name','word']).word.value_counts().reindex(['bull','bear'],fill_value=0)
    return g.rename_axis('word').reset_index(name='count')

result_2 = solution_2(files.copy())
display(result_2)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
print('All example checks passed')
```
