# 1378. Replace Employee ID With The Unique Identifier

## Problem Description

A company keeps its employee directory and its separate unique-ID assignments in two tables.

### Table Schema

**Employees**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| id | int | unique employee identifier (key) |
| name | varchar | employee name |

**EmployeeUNI**

| Column | Type | Meaning / constraint |
| --- | --- | --- |
| id | int | employee identifier |
| unique_id | int | assigned identifier |

### Task

The employee ID and unique ID pair identifies an assignment in EmployeeUNI. Return the assigned unique_id and employee name for all employees, preserving employees even when there is no matching assignment. In that case unique_id must be null. Return unique_id, name; any order is accepted.

## Example

### Input

**employees**

| id | name |
| --- | --- |
| 1 | A |
| 2 | B |
| 3 | C |

**employee_uni**

| id | unique_id |
| --- | --- |
| 1 | 10 |
| 3 | 30 |

### Expected Output

| unique_id | name |
| --- | --- |
| 10 | A |
| NULL | B |
| 30 | C |

### Explanation

Employees A and C have assigned identifiers 10 and 30. B has no matching assignment, so its unique_id is null; B must still appear.

```python
import pandas as pd
from pandas.testing import assert_frame_equal

employees = pd.DataFrame([[1,"A"],[2,"B"],[3,"C"]], columns=["id","name"])
employee_uni = pd.DataFrame([[1,10],[3,30]], columns=["id","unique_id"])
expected = pd.DataFrame([[10,"A"],[None,"B"],[30,"C"]], columns=["unique_id","name"])
display(employees)
display(employee_uni)
display(expected)
```

## Solutions

### Solution 1 — Left merge
保留左表所有員工；支援右表多筆匹配。

```python
def solution_1(employees, employee_uni):
    return employees.merge(employee_uni,on='id',how='left')[['unique_id','name']]

result_1 = solution_1(employees.copy(), employee_uni.copy())
display(result_1)
```

### Solution 2 — Index join
join 使用 index 作為匹配鍵。

```python
def solution_2(employees, employee_uni):
    return employees.join(employee_uni.set_index('id'),on='id',how='left')[['unique_id','name']]

result_2 = solution_2(employees.copy(), employee_uni.copy())
display(result_2)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
print('All example checks passed')
```
