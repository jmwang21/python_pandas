```python
import pandas as pd
from pandas.testing import assert_frame_equal

actor_director = pd.DataFrame([[1,1,1],[1,1,2],[1,1,3],[1,2,4],[2,1,5],[2,1,6]], columns=["actor_id","director_id","timestamp"])
expected = pd.DataFrame([[1,1]], columns=["actor_id","director_id"])
display(actor_director)
display(expected)
```

## Solutions

### Solution 1 — Group size
以兩個鍵計算 row count。

```python
def solution_1(actor_director):
    g = actor_director.groupby(['actor_id','director_id']).size().reset_index(name='n')
    return g.loc[g.n.ge(3),['actor_id','director_id']]

result_1 = solution_1(actor_director.copy())
display(result_1)
```

### Solution 2 — Pair value_counts
直接計算多欄位組合出現頻率。

```python
def solution_2(actor_director):
    g = actor_director[['actor_id','director_id']].value_counts()
    return g[g.ge(3)].reset_index(name='n')[['actor_id','director_id']]

result_2 = solution_2(actor_director.copy())
display(result_2)
```

### Solution 3 — Transform size
保留 row-level mask，可延伸篩選所有合作記錄。

```python
def solution_3(actor_director):
    g = actor_director.groupby(['actor_id','director_id']).timestamp.transform('size')
    return actor_director.loc[g.ge(3),['actor_id','director_id']].drop_duplicates()

result_3 = solution_3(actor_director.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
