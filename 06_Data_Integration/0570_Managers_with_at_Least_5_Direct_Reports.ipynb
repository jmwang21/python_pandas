## Problem
Employee(id,name,department,managerId): return names of managers having at least five direct reports. Count direct reports only, not descendants. Any order.

```python
import pandas as pd
from pandas.testing import assert_frame_equal

employee = pd.DataFrame([[1,"Manager","A",None],[2,"A","A",1],[3,"B","A",1],[4,"C","B",1],[5,"D","A",1],[6,"E","A",1],[7,"F","A",2]], columns=["id","name","department","managerId"])
expected = pd.DataFrame([["Manager"]], columns=["name"])
display(employee)
display(expected)
```

## Solutions

### Solution 1 — Value counts and isin
managerId 的缺值不算；用 id 回查姓名。

```python
def solution_1(employee):
    g = employee.managerId.value_counts()
    return employee.loc[employee.id.isin(g[g.ge(5)].index),['name']]

result_1 = solution_1(employee.copy())
display(result_1)
```

### Solution 2 — Aggregate and merge
先 groupby，後 self-style join。

```python
def solution_2(employee):
    g = employee.groupby('managerId').size().reset_index(name='n')
    return employee.merge(g.loc[g.n.ge(5)],left_on='id',right_on='managerId')[['name']]

result_2 = solution_2(employee.copy())
display(result_2)
```

### Solution 3 — Map counts
map 對應單一彙總值，不改變 row 數。

```python
def solution_3(employee):
    n = employee.id.map(employee.groupby('managerId').size()).fillna(0)
    return employee.loc[n.ge(5),['name']]

result_3 = solution_3(employee.copy())
display(result_3)
```

## Check all solutions
Rows are normalized only for the example comparison; preserve required ordering in the solutions themselves.

```python
def normalize(df):
    return df.sort_values(list(df.columns),na_position='last').reset_index(drop=True)
assert_frame_equal(normalize(result_1), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_2), normalize(expected), check_dtype=False)
assert_frame_equal(normalize(result_3), normalize(expected), check_dtype=False)
print('All example checks passed')
```
